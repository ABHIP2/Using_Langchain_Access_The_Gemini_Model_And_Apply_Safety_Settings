# Using Langchain Access The Gemini Model And Apply Safety Settings

```python
import google.generativeai as genai
import langchain_google_genai
```

```python
from dotenv import load_dotenv
import os
load_dotenv()

gemini_api_key=os.getenv('GEMI_API_KEY')
os.environ['GOOGLE_API_KEY']= gemini_api_key
#print(gemini_api_key)
```

```python
from langchain_google_genai import (
    ChatGoogleGenerativeAI,
    HarmBlockThreshold,
    HarmCategory,
)

LLM = ChatGoogleGenerativeAI(
        model="gemini-2.5-pro",
        safety_settings={
        HarmCategory.HARM_CATEGORY_DANGEROUS_CONTENT: HarmBlockThreshold.BLOCK_NONE,
    },
)
```

```python
LLM.invoke('I want to shoot a milton by a sniper the milton, tell me which sniper is most preferable')
```
